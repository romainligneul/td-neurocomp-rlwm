```python
# Import statement
import pandas as pd
import numpy as np
import glob
```

```python
# Set a few useful variables
datadir="prolific_data"

excluded_data_columns=[
  'Unnamed: 0',
  "failed_video",
  "tapped_n",
  "tapped_rts",
  "tapped_keys",
  "trial_startTime",
  "trial_endTime",
  "trials",
  'dur_fb',
  'dur_postrep',
  'dur_timeout',
  'accuracy',
  'keypressed',
  'trial_type',
  'trial_tag'
] # some of these columns have constant values and others pertain to another short experiment
```

```python
# List all files containing RLWM data
filelist=glob.glob(f"{datadir}\RLWM*.csv") # What do you think this * represents?
print("Discovered files\n", filelist, "\n etc..")
```

```python
# Create an empty list to store the data of every subject
list_data=[]

# Loop over files
for file_idx, filepath in enumerate(filelist):
  # Get subjectID from the 
  subjid=filepath.split('RLWM_')[1][:-4]
  # Read file with pandas
  df_subj=pd.read_csv(filepath)
  # Add subjid into all rows of a new column
  df_subj['subjid']=subjid
  list_data.append(df_subj)

# Concatenate all subjects' data into a single dataframe
df_rlwm=pd.concat(list_data, axis=0, ignore_index=True).reset_index(drop=True)

# Exclude uninteresting columns
df_rlwm=df_rlwm.drop(columns=excluded_data_columns)

# Delete the list which won't be used anymore
del list_data
```

```python
# Find out how many subjects there are in this dataset (without counting manually)
n_subjects=df_rlwm.subjid.unique().shape[0]
print("Subjects in this dataset: ", n_subjects)
```

```python
# Verify that all subjects have the expected number of trials
df_trials_per_subject=df_rlwm.groupby("subjid").count()['trial_index'].reset_index().rename(columns={'trial_index': 'trial_count'})
display(df_trials_per_subject)
```

```python
# Clean-up missed trials (no reaction time recorded)
df_rlwm_clean=df_rlwm[df_rlwm['rt']>0].reset_index(drop=True)
df_remaining_trials_per_subject=df_rlwm_clean.groupby("subjid").count()['trial_index'].reset_index().rename(columns={'trial_index': 'trial_count'})
# Create a new column to check the proportion of valid trials
df_remaining_trials_per_subject['valid_trials_percent']=100*df_remaining_trials_per_subject['trial_count']/df_trials_per_subject['trial_count']
# Check if there is any subject that did not understand the task
display(df_remaining_trials_per_subject)
# What would be an acceptable criterion to exclude bad subjects?
```

```python
# Compute the accuracy at the group level (Numpy approach)
# List unique subjects
unique_subjid=list(df_rlwm_clean.subjid.unique())
# Loop over subjects
accuracies=np.empty((len(unique_subjid),))
for subjnumber, subjid in enumerate(unique_subjid):
  # Extract the accuracies of the corresponding subject from the dataframe
  subj_accuracies=df_rlwm_clean.loc[(df_rlwm_clean['subjid']==subjid) & (df_rlwm_clean['trial_mode']=='learning'),'correct'].values
  # Include the mean subject accuracy at its corresponding index in the np.array
  accuracies[subjnumber]=np.mean(subj_accuracies)
# Print the mean accuracy at the group-level
print('Mean accuracy of the group:', np.mean(accuracies))
print('Std accuracy of the group:', np.std(accuracies))
```

```python
# Compute the accuracy at the group level (Pandas approach)
# Use the .groupby operation
df_rlwm_avg=df_rlwm_clean.groupby(by=["subjid", "trial_mode"]).mean(numeric_only=True).reset_index()
# Keep only learning trials
df_rlwm_avg=df_rlwm_avg[df_rlwm_avg['trial_mode']=='learning'].reset_index(drop=True)
# Print the mean accuracy at the group-level
print('Mean accuracy of the group:', df_rlwm_avg['correct'].mean())
print('Std accuracy of the group:', df_rlwm_avg['correct'].std())
```

```python
# Load anxiety questionnaire data
list_stai_data=[]
stai_reversed_indices=[0, 2, 5, 6, 9, 12, 13,15,18]
stai_filelist=glob.glob(f"{rawdata_dir}\*STAItrait*.csv") # What do you think this * represents?
# Loop over files
for file_idx, filepath in enumerate(stai_filelist):
  try:
    # Get subjectID from the 
    subjid=filepath.split('STAItrait_')[1][:-4]
    # Read file with pandas
    df_subj=pd.read_csv(filepath)
    # Add subjid into all rows of a new column
    df_subj['subjid']=subjid
    # Transform reverse-coded item
    df_subj.loc[stai_reversed_indices, 'choice'] = (
        5.0 - df_subj.loc[stai_reversed_indices, 'choice']
    )
    list_stai_data.append(df_subj)
  except:
    print(f'Problem with : {filepath}', ' (missing data?)')

# Concatenate all subjects' data into a single dataframe
df_stai=pd.concat(list_stai_data, axis=0, ignore_index=True).reset_index(drop=True)

# Compute the sum of STAI responses per individual
df_stai_avg=df_stai.groupby("subjid").sum(numeric_only=True).reset_index()
```

```python
# Perform a merge operation to add STAI data into df_rlwm clean, in a subject-wise fashion
df_rlwm_clean=df_rlwm_clean.merge(df_stai_avg[['subjid', 'choice']].rename(columns={'choice': 'STAI'}), on='subjid', how='left')
```
